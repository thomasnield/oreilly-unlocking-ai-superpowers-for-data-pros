# Tokenize words in a string


```python
import re

def break_up_words(s):
    return re.sub(r'[^\w]', '', s.lower()).split()

print(break_up_words("Hello, this is a sentence I am trying to tokenize."))
```

# Extracting Websites from a body of text

```python
import re
web_pattern = re.compile(r'(https?://)?(www\.)?([a-z0-9]+)(com|org|gov)', re.IGNORECASE)

urls = """
Here are a few websites below:

https://www.yawmanflight.com
http://microsoft.com
https://youtube.com
https://www.anaconda.com

These are non-commercial sites:
https://www.python.org
https://whitehouse.gov
"""

matches = web_pattern.finditer(urls)

for match in matches:
    print(match[0])
```
